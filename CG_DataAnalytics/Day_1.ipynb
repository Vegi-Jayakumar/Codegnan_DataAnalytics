```python
import pandas as pd
```

```python
#create pandas Series
data = [1,2,3,4,5,6,7,8,9,0]
s = pd.Series(data)
print(s)
```

```python
s=pd.Series([1,2,3,4,5,6,7,8,9,0],index=["a","b","c","d","e","f","g","h","i","j"])
print(s)
```

```python
dict_ = {'x':100,'y':200,'z':300}
s2 = pd.Series(dict_)
print(s2)
```

```python
#Accessing pandas Series
data = ['g','e','e','k','s','f','o','r','g','e','e','k','s']
s3 = pd.Series(data)
#first 5
print(s3[:5])
#first 3
print(s3[:3])
#last 4
print(s3[-4:])
#6-9
print(s3[5:9])
#reverse
print(s3[::-1])
#ascending
print(s3.sort_values)
#descending
print(s3[::-1].sort_values)
```

```python
#Updating pandas Series
data2 = {'a':10,'b':20,'c':30,'d':40}
s4 = pd.Series(data2)
print(s4)
#changing values
s4['c'] = 35
print(s4)
#Adding values
s4['e'] = 50
print(s4)
#Removing values
#Dynamic - removes only from the active cell and does not remove from the database
# s4.drop('b')
#static - removes from the database permanently
s4 = s4.drop('b')
print(s4)
```

```python
#Question 1
student_id = [101,102,103,104,105]
marks = [78,85,92,67,88]

#creating
s_marks = pd.Series(data = marks, index = student_id)
print("Creating")
print(s_marks)
#accessing
print("Accessing")
print(s_marks[103])
#updating
print("Updating")
s_marks[104] = 75
print(s_marks[104])
#deleting
print("deleting")
s_marks = s_marks.drop(102)
print(s_marks)
#Adding
print("Adding")
s_marks[106] = 90
print(s_marks)
```

```python
#Question 2
employee_id = [1001,1002,1003,1004,1005]
salary = [35000,42000,38000,50000,45000]

#Creating
print("Creating")
e_sal = pd.Series(data = salary,index = employee_id)
print(e_sal)
#Updating
print("Updating")
e_sal[1003] = 40000
print(e_sal[1003])
#deleting
print("Deleting")
e_sal = e_sal.drop(1002)
print(e_sal)
#Adding
print("Adding")
e_sal[1006] = 55000
print(e_sal)
#Accessing
print("Accessing")
print(e_sal[:3])
```

```python
#Question 3
product_id = ['p101','p102','p103','p104','p105']
price = [500,750,1200,900,1500]

#creating
p_price = pd.Series(data = price, index = product_id)
print("Creating")
print(p_price)
#accessing
print("Accessing")
print(p_price['p103'])
#updating
print("Updating")
p_price['p102'] = 800
print(p_price['p102'])
#deleting
print("deleting")
p_price = p_price.drop('p104')
print(p_price)
#Adding
print("Adding")
p_price['p106'] = 2000
print(p_price)
```
