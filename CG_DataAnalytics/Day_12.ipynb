```python
import numpy as np
```

```python
#Random Module
```

```python
# Generates values between 0 - 1
arr = np.random.random(5)
print(arr)
```

```python
arr1 = (10*np.random.random(5)).astype(int)
print(arr1)
```

```python
# Generates float numbers between the given range
arr2 = np.random.uniform(10,50,5)
print(arr2)
```

```python
# generates values whose mean = 0 and std div = 1
arr3 = np.random.randn(5)
print(arr3)
```

```python
# generates values from the given list
arr4 = np.random.choice(["Rock 🪨", "Paper 📄", "Scissors ✂"])
print(arr4)
arr5 = np.random.choice(["Heads 🗿", "Tails 🐒"], 2)
print(arr5)
```

```python
# Assign probabilities to the choices
arr6 = [10,20,30]
result = np.random.choice(arr6,size = 2,p=[0.2,0.5,0.3])
print(result)
```

```python
# Freeze the already generated values
arr7 = np.random.random(5)
np.random.seed(10)
print(arr7)
```

```python
# Generate random integer values under the given range
arr8 = np.random.randint(10,50,4)
print(arr8)
```

```python
# Randomly changes the order of the array given permanently
arr9 = np.array([10,20,30,40,50])
np.random.shuffle(arr9)
print(arr9)
```

```python
# Randomly changes the order of the array given temperorily
arr10 = np.array([10,20,30,40,50])
test = np.random.permutation(arr10)
print(f"Original: {arr10}\nPermutated: {test}")
```
