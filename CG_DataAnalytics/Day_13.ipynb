```python
import numpy as np
```

```python
# Aggregate Function
```

```python
arr = np.array([10,20,30,40,50])

#Sum
print(np.sum(arr))
```

```python
#Min
print(np.min(arr))
```

```python
#Max
print(np.max(arr))
```

```python
#Mean
print(np.mean(arr))
```

```python
#Variance
print(np.var(arr))
```

```python
#Standard Deviation
print(np.std(arr))
```

```python
#Median
print(np.median(arr))
```

```python
#Percentile
print(np.percentile(arr,0))
print(np.percentile(arr,25))
print(np.percentile(arr,50))
print(np.percentile(arr,75))
print(np.percentile(arr,100))
```

```python
#Quantile
print(np.quantile(arr, 0.0))
print(np.quantile(arr, 0.25))
print(np.quantile(arr, 0.50))
print(np.quantile(arr, 0.75))
print(np.quantile(arr, 1.0))
```

```python
#Argmax
print(np.argmax(arr))
```

```python
#Argmin
print(np.argmin(arr))
```
