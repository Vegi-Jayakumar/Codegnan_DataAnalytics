```python
# Aggregate function
import numpy as np
```

```python
# log()/ log2()/ log10()
arr = np.array([10,100,1000], int)
print("Base 10")
print(np.log10(arr))
print("Base 2")
print(np.log2(arr))
print("Base e")
print(np.log(arr))
```

```python
#abs values
arr1 = np.array([-10,-5,0,5,10], int)
print(np.abs(arr1))
```

```python
# ascending sorting numbers
arr2 = np.sort([50,40,10,20,30])
print(arr2)
```

```python
# ascending sorting strings
names = np.array(["Ravi","Arun","Priya","Kiran"], str)
result = np.sort(names)
print(result)
```

```python
# descending sorting strings
names = np.array(["Ravi","Arun","Priya","Kiran"], str)
result_d = np.sort(names)[::-1]
print(result_d)
```
