```python
import numpy as np
```

```python
arr = np.array([
    [1,2,3],
    [4,5,6]
])
print(arr)
print()
ary = np.array([
    [10,20,30],
    [40,50,60]
])
print(ary)
```

```python
# Transpose
result = arr.T
print(result)
print()
result1 = arr.transpose()
print(result1)
```

```python
# Ravel
result2 = arr.ravel()
print(result2)
```

```python
# Insertion in 2D array
arr1 = np.array([10,20,30,40])
new_arr1 = np.insert(arr1, 2, 100)
print(new_arr1)
new_arr = np.insert(arr1, 3, [200, 500])
print(new_arr)
```

```python
# Deletion in 2D array
new_arr = np.delete(arr1, 3)
print(new_arr)
```

```python
# Merging 2D arrays
result = np.concatenate((arr,ary), axis=1)
print(result)
print()
result1 = np.concatenate((arr,ary),axis=0)
print(result1)
```

```python
# Stacking 1D arrays
arr2 = np.array([50,60,70,80])
result3 = np.stack((arr1, arr2), axis=0)
print(result3)
print()
result4 = np.stack((arr1,arr2),axis=1)
print(result4)
```

```python
# Stacking 2D arrays
result5 = np.stack((arr,ary),axis=0)
print(result5)
print()
result6 = np.stack((arr,ary),axis=1)
print(result6)
```

```python
# Stacking 3D arrays
arr3 = np.arange(1,28)
arr3 = arr3.reshape(3,3,3)
arr4 = np.arange(3,30)
arr4 = arr4.reshape(3,3,3)
result7 = np.stack((arr3, arr4), axis=0)
print(result7)
print()
result8 = np.stack((arr3, arr4), axis=1)
print(result8)
```
