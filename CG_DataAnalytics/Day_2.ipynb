A DataFrame is a 2-dimensional labeled data structure in pandas.
Eg: Excel sheet, SQL table
We can create DataFrame using nested list, nested tuple, set, dictionary.

```python
import pandas as pd
```

```python
#Using List
data = [
    ["Ravi",22,30000],
    ["Priya",24,40000],
    ["Arun",21,28000]
]

df = pd.DataFrame(data,columns=["Name","Age","Salary"])
print(df)
```

```python
#Using Tuple
data = (
    ("Ravi",22,30000),
    ("Priya",24,40000),
    ("Arun",21,28000)
)

df = pd.DataFrame(data,columns=("Name","Age","Salary"))
print(df)
```

```python
#Using Dictionary
data = {
    "Name":["Ravi","Priya","Arun"],
    "Age":[22,24,21],
    "Salary":[30000,40000,28000]
}

df = pd.DataFrame(data)
print(df)
```

```python
#create a DataFrame for sales and profit of every month
#for first six months of the year

data = {
    'Month' : ['January','February','March','April','May','June'],
    'Sale' : [3000000,5000000,2000000,8000000,3500000,4000000],
    'Profit' : [300000,500000,200000,800000,350000,400000]
}

df = pd.DataFrame(data)
print(df)
```

```python
#Using Set
my_set = {100,200,300,400}

df = pd.DataFrame(list(my_set),columns=['Values'])
print(df)
```

Converting Series to DataFrames:
for single series we use to_frame() method.
for multiple series we use a dictionary for conversion.

```python
#Single Series
#Assigning/Rename Columns
s = pd.Series([1,3,5,7,9],name = "Values")
df = s.to_frame()
df.columns = ["Digits"]
print(df)
```

```python
#Multiple Series
#Assigning/Rename multiple Columns
s1 = pd.Series([2,4,6,8,10])
s2 = pd.Series([1,3,5,7,9])
d = {
    "Even":s1,
    "Odd":s2
}
df = pd.DataFrame(d)
df.columns = ["Even Nums","Odd Nums"]
print(df)
```

```python
#Rename Column names permanently
df.rename(
    columns={
        "Even Nums":"EVEN",
        "Odd Nums":"ODD"
    },
    inplace=True
)
print(df)
```

```python
#Viewing dataframe
#head() --> gives top 5 rows by default (we can customize)
#tail() --> gives bottom 5 rows by default (we can customize)
print("from top")
print(df.head(2))
print("\nfrom bottom")
print(df.tail(2))
```

```python
#len() method gives the number of rows in a dataframes
#len() method gives number of values in a Series
print(len(df))

#shape method gives the dimensions of a table
print(df.shape)

#values method to show the values
print(df.values)

#index method to show the indexs of each element in a dataframe
print(df.index)
```
