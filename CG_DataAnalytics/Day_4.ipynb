```python
import pandas as pd
```

```python
#Manipulating data
sales = pd.Series(
    [5000,7000,None,9000,8000,10000,None],
    index = ['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday'],
    name = "Sales"
)
print(sales)
```

```python
#Aggregate functions
print(f"min: {sales.min()}")
print(f"max: {sales.max()}")
print(f"sum: {sales.sum()}")
print(f"Count: {sales.count()}")
print(f"median: {sales.median()}")
print(f"mean: {sales.mean()}")
print(f"variance: {sales.var()}")
print(f"Standard Deviation: {sales.std()}")
print(f"DataType: {sales.dtype}")
```

```python
#Updating
sales["Tuesday"] = 9500

#Adding
sales["January"] = 4*sales.sum()

#Droping
sales = sales.drop("Thursday")

print(sales)
```

```python
#updating all the sales
print(sales + 2000)
```

```python
s1 = pd.Series([1,2,3,4,5])
s2 = pd.Series([101,102,103,104,105])

#Concat
combined = pd.concat([s1,s2],ignore_index=True)    #Use ignore_index to continue the index rather than disjoint index
print(combined)
```

```python
data = {
    "Name": ["Ravi","Priya","Arun","Sneha","Kiran","Anjali","Vikram","Meena"],
    "Age": [22,24,21,25,23,26,20,27],
    "City": ["Hyderabad","Chennai","Banglore","Hyderabad","Mumbai","Chennai","Banglore","Mumbai"],
    "Salary": [30000,40000,28000,50000,35000,45000,25000,55000]
}

df = pd.DataFrame(data, index = [101,102,103,104,105,106,107,108])
print(df)
```

```python
#Aggregation functions
print("Aggregate functions of Age")
print(f"min: {df["Age"].min()}")
print(f"max: {df["Age"].max()}")
print(f"sum: {df["Age"].sum()}")
print(f"Count: {df["Age"].count()}")
print(f"median: {df["Age"].median()}")
print(f"mean: {df["Age"].mean()}")
print(f"variance: {df["Age"].var()}")
print(f"Standard Deviation: {df["Age"].std()}")
print(f"DataType: {df["Age"].dtype}")
print('-'*30)
print("Aggregate functions of Salary")
print(f"min: {df["Salary"].min()}")
print(f"max: {df["Salary"].max()}")
print(f"sum: {df["Salary"].sum()}")
print(f"Count: {df["Salary"].count()}")
print(f"median: {df["Salary"].median()}")
print(f"mean: {df["Salary"].mean()}")
print(f"variance: {df["Salary"].var()}")
print(f"Standard Deviation: {df["Salary"].std()}")
print(f"DataType: {df["Salary"].dtype}")
```

```python
#Adding new column
df["Bonus"] = df["Salary"]*0.10
df["New_Salary"] = df["Salary"] + df["Bonus"]
print(df)
```

```python
#Updating a Column
df["Age"] = df["Age"] + 1
df["New_Salary"] = df["New_Salary"] + 500
print(df)
```

```python
#Updating a Specific Value
df.loc[df["Name"] == "Priya", "City"] = "Pune"
df.loc[df["City"] == "Banglore", "City"] = "Hyderabad"
df.loc[df["City"] == "Mumbai", "Salary"] = df["Salary"]*1.05
df.loc[df["Salary"] < 45000, "Salary"] = 60000.0
print(df)
```

```python
#Deleting a column
#axis = 1 means column, axis = 0 means row
df.drop("New_Salary", axis = 1, inplace=True)
print(df)
```

```python
#Use columns =
df.drop(columns = ["Bonus"], inplace = True)
print(df)
```
