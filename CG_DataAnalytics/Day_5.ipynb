```python
import pandas as pd
```

to read csv files: df = pd.read_csv("sales.csv")

We can Identify Missing Values using isna() or isnull().
To get the count of missing values using isna().sum() or isnull().sum().
We can fill the values in missing values using fillna().
We can remove the missing values using dropna().

```python
#Cleaning Data
sales = pd.Series(
    [5000,7000,None,9000,8000,10000,None],
    index = ['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday'],
    name = "Sales"
)
print(sales)
```

```python
#Detecting Missing Values
print(sales.isna())
```

```python
#Counting the missing values
print(sales.isna().sum())
```

```python
#Removing missing values
print(sales.dropna())
```

```python
#Filling the missing values
print(sales)
print("-"*30)
print(sales.fillna(0))
print("-"*30)
print(sales.fillna("No Data"))
print("-"*30)
print(sales.fillna(sales.mean()))
```

```python
employees = {
  "EmpID": [101, 102, 103, 104],
  "Name": ["Asha", None, "Meena", "Sai"],
  "Department": ["Sales","IT","HR",None],
  "Salary": [45000, 60000, None, 89000]
}
I = [901, 902, 903, 904]
df_emp = pd.DataFrame(employees, index = I)
print(df_emp)
```

```python
#Find and count the missing values
print(df_emp.isna())
print("\ncount of missing values")
print(df_emp.isna().sum())
```

```python
#Filling different missing values from different columns
print(df_emp.fillna(
    {
        "Name":"Unknown",
        "Department":0,
        "Salary":0
    }
))
```

```python
#Filling Missing salary with max value
print(df_emp["Salary"].fillna(df_emp["Salary"].max()))
```

```python
#Filling Missing salary with min value
print(df_emp["Salary"].fillna(df_emp["Salary"].min()))
```

```python
#Filling Missing salary with mean value
print(df_emp["Salary"].fillna(df_emp["Salary"].mean()))
```

```python
#Filling Missing salary with variance value
print(df_emp["Salary"].fillna(df_emp["Salary"].var()))
```

```python
#Filling Missing salary with standard deviation value
print(df_emp["Salary"].fillna(df_emp["Salary"].std()))
```

```python
#Filling Missing salary with median value
print(df_emp["Salary"].fillna(df_emp["Salary"].median()))
```
