```python
import pandas as pd
```

```python
#Duplicate data handling
data = {
    "EmpID" : [101,102,103,104,105,106,107,108,109,110,110],
    "Name" : ["Ravi","PRIYA","Arun","sneha","Kiran","ANJALI","Vikram","meena","RAHUL","Sneha","Sneha"],
    "City" : ["Hyderabad","hyderabad","CHENNAI","Chennai","Banglore","BANGLORE",
              "Mumbai","mumbai","Delhi","Hyderabad","Hyderabad"],
    "Salary" : [30000,40000,3500,4500,32000,50000,38000,42000,36000,45000,45000]
}
df = pd.DataFrame(data)
print(df)
```

```python
#Find duplicate rows
print(df.duplicated())
print()

#Count of duplicate rows
print(f"number of duplicate rows: {df.duplicated().sum()}")
print()

#Display duplicate rows
print(df[df.duplicated()])
print()

#Display the occurances of duplicate rows
print(df[df.duplicated(keep = False)])
```

```python
#Display Individual duplicate columns
duplicate_rows = df[df.duplicated()]
total_Salary = duplicate_rows["Salary"].sum()
total_Name = duplicate_rows["Name"].sum()
total = duplicate_rows[["Name","City"]].sum()
print(total_Salary)
print()
print(total_Name)
print()
print(total)
print()
print(df["EmpID"].duplicated())
```

```python
#Drop Duplicates
#It removes the duplicate while keeping the original
df = df.drop_duplicates() # or drop_duplicates(keep = "first")
print(df)
```

```python
# If you want to keep the duplicate and remove the original
df = df.drop_duplicates(keep = "last")
print(df)
```

```python
#Remove all the duplicate occurances including the original
df = df.drop_duplicates(keep = False)
print(df)
```
