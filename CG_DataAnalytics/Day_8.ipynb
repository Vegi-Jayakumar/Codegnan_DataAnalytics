```python
import pandas as pd
```

```python
data = {
   "VisitID": [501, 502, 503, 504, 505, 506, 507, 508, 509, 510, 510, 511],
   "Patient": ["Rahul", "Anita", "Suresh", "Priya", "Kiran", "Anjali", "Ravi", "Meena", "Arun", "Priya", "Priya", "Kiran"],
   "Department": ["Cardiology", "Neurology", "Orthopedic", "General", "Dental", "Cardiology", "Neurology", "General", "Dental", "Cardiology", "Cardiology", "Dental"],
   "BillAmount": [5000, 7500, 4500, 2000, 1500, 6000, 7000, 2500, 1800, 6000,6000, 1500]}

df = pd.DataFrame(data)
print(df)
```

```python
# 1
print(df[df.duplicated()])
```

```python
# 2
print(df.duplicated().sum())
```

```python
# 3
print(df[df.duplicated(keep = False)])
```

```python
# 4
amount = df[df.duplicated(keep=False)]
total = amount["BillAmount"].sum()
print(total)
```

```python
# 5
duplicates = df[df.duplicated(keep=False)]
print(duplicates["VisitID"])
```

```python
# 6
print(df[df["Patient"].duplicated(keep=False)])
```

```python
# 7
print(df.drop_duplicates())
```

```python
# 8
print(df["VisitID"].drop_duplicates())
```

```python
# 9
print(df[["Patient","Department"]].drop_duplicates())
```
