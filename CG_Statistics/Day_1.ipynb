Case Study: Student Performance System

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
students = pd.DataFrame({
    "Student_id":["CGVI0185","CGVI0206","CGVI0213","CGVI0184","CGVI0207"],
    "Course":["DA","DA","DA","PFS","PFS"],
    "Hours":[5,8,3,10,6],
    "Score":[85,70,65,55,40],
    "Satisfaction":["Very Satisfied","Satisfied","Neutral","Satisfied","Neutral"]
})
students
```

```python
print(f"Shape: {students.shape}")
print(f"Table type: {type(students)}")
print(f"Data type:\n{students.dtypes}")
print(f"Size: {students.size}")
print(f"Dimensions: {students.ndim}")
print(f"column type: {type(students.values)}")
print(f"Missing Values:\n{students.isna().sum()}")
```

```python
#Counting Categories and Summaris Numerical Columns
print("Students per Course Count:\n",students["Course"].value_counts())
print(students[["Hours","Score"]].describe())   #Descriptive Stats information
#Observe Students Proportion
print("Percentage of DA Students",(students["Course"] == "DA").mean())
print("Percentage of PFS Students",(students["Course"] == "PFS").mean())
```

```python
#Python Central tendency and Interpretation
values = np.array([12,15,15,18,20,22,22,22,25,90])
mean_value = np.mean(values)
median_value = np.median(values)
mode_value = pd.Series(values).mode().tolist()
print(f"Mean: {mean_value}\nMedian: {median_value}\nMode: {mode_value}")
print(f"Sorted Values {np.sort(values)}")
```
